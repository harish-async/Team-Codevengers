```python
import time
import joblib
import pandas as pd
import xgboost as xgb
```

```python
df =  pd.read_parquet("../Dataset/final-dataset/dataset-with-features.parquet")
df_model = df.dropna(subset=["target_log_flux_1h"]).copy()

train_df = df_model.loc["2011":"2018"]

TARGET = 'target_log_flux_1h'

ignore_col = [TARGET, 'E2W_COR_FLUX']
feature_col = [col for col in train_df.columns if col not in ignore_col]

X_train =  train_df[feature_col]
y_train_log = train_df[TARGET]
```

```python
final_model = xgb.XGBRegressor(
    n_estimators = 350,
    learning_rate = 0.03,
    max_depth = 6,
    random_state = 42,
    n_jobs = -1,
    subsample = 0.8,
    reg_alpha = 0.1,
    reg_lambda = 2.0,
    colsample_bytree = 0.8,
    min_child_weight = 5
)
```

```python
print("Model Training Started...")
start_time = time.time()
final_model.fit(X_train, y_train_log)
training_time = time.time() - start_time
print(f"Model Training completed in {training_time} seconds.")
```

```python
model_filename = "surya-kavach.joblib"
file_path = f"Model/{model_filename}"
joblib.dump(final_model, file_path)
print(f"Final model saved successfuly with name: {model_filename}")
```
